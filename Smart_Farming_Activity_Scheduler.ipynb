```python

# Smart Farming Activity Scheduler
# Using Graph and Topological Sorting

# Graph: Task -> Tasks that depend on it
graph = {}

def add_task():
    task = input("Enter farming task name: ").strip()

    if not task:
        print("Task name cannot be empty!")
    elif task in graph:
        print("Task already exists!")
    else:
        graph[task] = []
        print("Task added successfully!")


def add_dependency():
    if not graph:
        print("Please add tasks first!")
        return

    print("\nAvailable tasks:", list(graph.keys()))

    first = input("Enter prerequisite task: ").strip()
    second = input("Enter task that depends on it: ").strip()

    if first not in graph or second not in graph:
        print("Please enter existing task names!")
    elif first == second:
        print("A task cannot depend on itself!")
    elif second in graph[first]:
        print("Dependency already exists!")
    else:
        graph[first].append(second)
        print("Dependency added successfully!")


def show_tasks():
    if not graph:
        print("No tasks added yet!")
        return

    print("\nFarming Tasks and Dependencies:")
    for task in graph:
        print(task, "->", graph[task])


def create_schedule():
    if not graph:
        print("Please add farming tasks first!")
        return

    # Count prerequisites for each task
    indegree = {task: 0 for task in graph}

    for task in graph:
        for next_task in graph[task]:
            indegree[next_task] += 1

    # Tasks with no prerequisites
    queue = [
        task for task in graph
        if indegree[task] == 0
    ]

    schedule = []

    # Topological Sorting
    while queue:
        task = queue.pop(0)
        schedule.append(task)

        for next_task in graph[task]:
            indegree[next_task] -= 1

            if indegree[next_task] == 0:
                queue.append(next_task)

    # Check for circular dependencies
    if len(schedule) != len(graph):
        print("\nCircular dependency found!")
        print("A valid schedule cannot be created.")
    else:
        print("\n--- Smart Farming Schedule ---")
        for i, task in enumerate(schedule, start=1):
            print(f"{i}. {task}")


# Main Menu
while True:
    print("\n===== SMART FARMING SCHEDULER =====")
    print("1. Add Farming Task")
    print("2. Add Task Dependency")
    print("3. View Tasks")
    print("4. Generate Schedule")
    print("5. Exit")

    choice = input("Enter your choice: ").strip()

    if choice == "1":
        add_task()
    elif choice == "2":
        add_dependency()
    elif choice == "3":
        show_tasks()
    elif choice == "4":
        create_schedule()
    elif choice == "5":
        print("Thank you for using Smart Farming Scheduler!")
        break
    else:
        print("Invalid choice! Please try again.")
```
